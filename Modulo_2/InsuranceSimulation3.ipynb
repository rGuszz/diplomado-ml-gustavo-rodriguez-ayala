```python
#Simular densidades asociadas a seguros
#Densidades discreta
#Bernoulli
np.random.seed(123)
prob_success = 0.05
num_samples = 1000  #Número de muestras

bernoulli_samples = stats.bernoulli.rvs(p=prob_success, size=num_samples)

print(f"Primeras 20 Muestras Bernoulli (n={num_samples}, p={prob_success}): {bernoulli_samples[:20]}")
```

```python
bernoulli_samples.mean()
```

```python
plt.figure(figsize=(6, 4))
plt.hist(bernoulli_samples)
plt.title(f'Bernoulli Distribution Samples (n={num_samples}, p={prob_success})')
plt.xlabel('Número de éxitos')
plt.ylabel('Masa de probabilidad')
plt.grid(axis='y')
plt.show()
```

```python
sns.histplot(bernoulli_samples)
```

```python
# Binomial
np.random.seed(123)
n = 10 # número de ensayos bernoulli
p = 0.3 # probabilidad de éxito en cada ensayo

rvs = binom.rvs(n, p, size=1000)

# Cálculo de la función de masa de probabilidad
x = range(n + 1) #Rango de valores posibles
pmf = binom.pmf(x, n, p)

# Plot
plt.bar(x, pmf)
plt.xlabel('Número de éxitos')
plt.ylabel('Probabilidad')
plt.title(f'Simulación: Distribución Binomial (n={n}, p={p})')
plt.show()
```

```python
sns.histplot(rvs, discrete=True)
plt.title(f'Simulación: Distribución Binomial (n={n}, p={p})')
plt.xlabel('Número de éxitos')
plt.ylabel('Probabilidad')
plt.show()
```

```python
#Poisson
np.random.seed(123)
mu = 5 # Tasa media de ocurrencia (Intensidad)
x = np.arange(0, 20, 1) # Rango de valores

# Función de masa de probabilidad

y = poisson.pmf(x, mu)
# Gráfica
plt.plot(x, y, marker='o')
plt.title('Distribución Poisson (λ = 5)')
plt.xlabel('Número de eventos')
plt.ylabel('Probabilidad')
plt.grid()
plt.show()
```

```python
plt.bar(x, y)
plt.xlabel('Número de eventos')
plt.ylabel('Probabilidad')
plt.title(f'Simulación: Distribución Poisson (λ = {mu})')
plt.show()
```

```python
#Distribuciones propias de riesgo (Distribuciones con soporte positivo)
#Exponencial
np.random.seed(123)
exp_rv=expon.rvs(loc=0, scale=2.3, size=10000)
```

```python
plt.hist(exp_rv, bins=50,density=True,label='Sim',color='b', edgecolor='black')
x = np.arange(0,15,0.01)
plt.plot(x,expon.pdf(x,0,2.3),label=f'exponencial({2.3}) PDF')
plt.legend()
plt.show()
```

```python
#Distribución gamma
np.random.seed(123)
gamma_rv=gamma.rvs(2,0,50,5000)
```

```python
plt.hist(gamma_rv, bins=50,density=True,label='Sim',color='b', edgecolor='black')
x = np.arange(0,1000,0.1)
plt.plot(x,gamma.pdf(x,2,0,50),label=f'gamma({2},{100}) PDF')
plt.legend()
plt.show()
```

```python
#Log-normal
np.random.seed(123)
mu, sigma = 1.0, 0.5 
rng = np.random.default_rng()
logn_rv = rng.lognormal(mean=mu, sigma=sigma, size=1000)
```

```python
plt.figure(figsize=(8, 5))
plt.hist(logn_rv, bins=50, density=True, alpha=0.6, color='b', edgecolor='black')

x = np.linspace(min(logn_rv), max(logn_rv), 1000)
pdf = lognorm.pdf(x, s=sigma, scale=np.exp(mu))
plt.plot(x, pdf, 'r', linewidth=2, label="Log-Normal PDF")

plt.title("Log-Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Density")
plt.legend()
plt.show()
```

```python
#Weibull
shape = 2.0  # k
scale = 1.5  # λ
loc = 0.0    # θ

dist = weibull_min(shape, loc, scale)

print(f"Distribution parameters: shape={shape}, scale={scale}, loc={loc}")
```

```python
np.random.seed(123)
weibull_rv = dist.rvs(size=5000)
```

```python
x = np.linspace(min(weibull_rv), max(weibull_rv), 100)
pdf = dist.pdf(x)

plt.figure(figsize=(10, 6))
plt.hist(weibull_rv, bins=50, density=True, alpha=0.6, color='b', edgecolor='black')
plt.plot(x, pdf, 'r-', lw=2, label='PDF')
plt.title('Weibull Distribution PDF')
plt.xlabel('x')
plt.ylabel('Probability Density')
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Pareto
shape = 3.0   # Shape parameter (alpha > 0)
scale = 1.0   # Scale parameter (xm > 0)
size = 10000  # Number of samples

pareto_rv = (np.random.pareto(shape, size) + 1) * scale

x = np.linspace(min(pareto_rv), max(pareto_rv), 100)  
pdf_values = pareto.pdf(x, shape, scale=scale)

plt.figure(figsize=(10, 6))
plt.hist(pareto_rv, bins=50, density=True, alpha=0.6, color='b', edgecolor='black')
plt.plot(x, pdf_values, 'r-', lw=2, label='PDF')
plt.title('Pareto Distribution PDF')
plt.xlabel('x')
plt.ylabel('Probability Density')
plt.legend()
plt.grid(True)
plt.show()
```

```python
c = 5.5   # shape parameter 1
d = 6.0   # shape parameter 2
loc = 0   # location parameter
scale = 10 # scale parameter

np.random.seed(123)  
burr_rv = burr.rvs(c, d, loc=loc, scale=scale, size=5000)

x = np.linspace(min(burr_rv), max(burr_rv), 100)
pdf_values = burr.pdf(x, c, d, loc=loc, scale=scale)
plt.figure(figsize=(10, 10))
plt.plot(x, pdf_values, 'r-', lw=2, label='PDF')
plt.hist(burr_rv, bins=30, density=True, alpha=0.5, color='b', label='samples',edgecolor='black')
plt.title('Burr Distribution - PDF')
plt.xlabel('x')
plt.ylabel('Density')
plt.legend()

```

```python
#¿Y qué hacemos una vez ajustada esta distribución a nuestros datos?
#VaR empírico
np.percentile(monto.monto,95)
```

```python
#Percentil teórico con el modelo burr
c = 1.505   # shape parameter 1
d = 1.422   # shape parameter 2
loc = 0   # location parameter
scale = 2725.107 # scale parameter

VaR_burr=burr.ppf(0.95,c, d, loc=loc, scale=scale)
print(f"VaR Burr: {VaR_burr}")
```

```python
#VaR por simulación

VaR_burr = []

for i in range(5000):
    VaR_burr.append(
        np.percentile(burr.rvs(c, d, loc=loc, scale=scale,size=500),95)
    )

print(VaR_burr[:5])
```

```python
#Estimador puntual

np.mean(VaR_burr)

# 95% Confidence Interval for our VaR estimate
ci_lower = np.percentile(VaR_burr, 2.5)

ci_upper = np.percentile(VaR_burr, 97.5) 

print(f"95% Confidence Interval for 95% VaR: [{ci_lower:.4f}, {ci_upper:.4f}]")
```

```python
# VaR por Bootstrap

VaR_boot = []

for i in range(5000):
    VaR_boot.append(
        np.percentile(monto.sample(frac=1,replace=True),95)
    )

print(VaR_boot[:5])
```

```python
# 95% boot Confidence Interval for our VaR estimate
ci_lower = np.percentile(VaR_boot, 2.5)

ci_upper = np.percentile(VaR_boot, 97.5) 

print(f"95% Bootstrap Confidence Interval for 95% VaR: [{ci_lower:.4f}, {ci_upper:.4f}]")
```
